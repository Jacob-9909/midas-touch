# 로컬 데이터 확인용 노트북
이 노트북은 로컬에 저장된 `data/base_personas.csv` (기본 데이터)와 LLM이 생성한 `data/augmented_personas.csv` (금융 정보가 추가된 데이터)를 비교하고 확인하는 용도입니다.

```python
import pandas as pd

# 1. 로컬 기본 데이터셋 로드
try:
    df_base = pd.read_csv("data/base_personas.csv")
    print(f"[기본 데이터] 로드 완료! 크기: {df_base.shape}")
    display(df_base.head(3))
except FileNotFoundError:
    print("data/base_personas.csv 파일이 없습니다. 터미널에서 `uv run export_csv.py`를 먼저 실행해주세요.")
```

```python
# 2. LLM이 생성/병합한 최종 데이터셋 로드
try:
    df_augmented = pd.read_csv("data/augmented_personas.csv")
    print(f"\n[증강된 데이터] 로드 완료! 크기: {df_augmented.shape}")
    
    # 새롭게 추가된 금융 컬럼 14개 모두 살펴보기
    finance_cols = [
        'total_amount', 'has_stock', 'has_bond', 'has_deposit', 'has_real_estate', 
        'specific_items', 'monthly_income', 'monthly_investable', 'aggressiveness', 
        'preferred_asset', 'financial_literacy', 'target_return_percent', 
        'investable_period_months', 'requires_liquidity'
    ]
    display(df_augmented[finance_cols].head(5))
except FileNotFoundError:
    print("data/augmented_personas.csv 파일이 없습니다. 터미널에서 `uv run generate_finance_data.py`를 먼저 실행해주세요.")
```
